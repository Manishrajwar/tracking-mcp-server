1. Install uv

2. create a project folder named fastmcp-demo-server

3. open the folder in vs code 

4. open terminal 

5. execute the command uv init

6. uv add fastmcp 

7. fastmcp version

8. create a basic server

9. test the server - uv run fastmcp dev main.py //  uv run fastmcp dev inspector main.py

10. Run the server - uv run fastmcp run main.py 

11. Add the server to claude desktop - uv run fastmcp install claude-desktop main.py
